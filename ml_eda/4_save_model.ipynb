## Сохранение модели в файл

```python
import joblib
import torch

model_data = {
    "count_vectorizer": count_vectorizer,
    "log_P_good": log_P_good,
    "log_P_sad": log_P_sad,
    "log_P_word_good": log_P_word_good,
    "log_P_word_sad": log_P_word_sad,
}

for key in model_data:
    print(f"{key}: {model_data[key]}")

print()
joblib.dump(model_data, "sentiment_naive_bayes.pkl")
```

## Импорт модели

```python
import joblib
import torch

model_data = joblib.load("sentiment_naive_bayes.pkl")

count_vectorizer = model_data["count_vectorizer"]
log_P_good = model_data["log_P_good"]
log_P_sad = model_data["log_P_sad"]
log_P_word_good = model_data["log_P_word_good"]
log_P_word_sad = model_data["log_P_word_sad"]
```
